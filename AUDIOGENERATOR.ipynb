```python
import gradio as gr
import numpy as np
import soundfile as sf
import torch
from qwen_tts import Qwen3TTSModel

print("Loading model onto Google GPU...")
model = Qwen3TTSModel.from_pretrained(
    "Qwen/Qwen3-TTS-12Hz-1.7B-VoiceDesign",
    device_map="cuda:0",
    dtype=torch.bfloat16,
)

def generate_narrator_audio(script_text, voice_description):
    wavs, sr = model.generate_voice_design(
        text=script_text, language="English", instruct=voice_description
    )
    sf.write("documentary_narration.wav", wavs[0], sr)
    return (sr, wavs[0]), "documentary_narration.wav"

interface = gr.Interface(
    fn=generate_narrator_audio,
    inputs=[
        gr.Textbox(
            lines=8,
            label="Script Text",
            placeholder="Paste your script here...",
        ),
        gr.Textbox(
            lines=3,
            label="Voice Tone Instruction",
            value="A calm, authoritative documentary narrator with deliberate pacing, slight dry wit, and a steady tone.",
        ),
    ],
    outputs=[
        gr.Audio(label="Preview Audio"),
        gr.File(label="Download WAV File"),
    ],
    title="Qwen3-TTS Studio Narration Generator",
)

interface.launch(share=True)
```
